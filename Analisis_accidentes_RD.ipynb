```python
# EDA
df_lesion_hora_mes


print("Info: \n", df_lesion_hora_mes.info())
print("Head: \n", df_lesion_hora_mes.head())
print("Tail: \n", df_lesion_hora_mes.tail())
print("Describe: \n", df_lesion_hora_mes.describe())
print("Duplicados:\n", df_lesion_hora_mes.duplicated().sum())
#Cambiar nombre a columna cantidad de muerte,nombrada asi por error
df_lesion_hora_mes.rename(columns={"Cantidad_muertes":"Cantidad_lesionados"}, inplace=True)
df_lesion_hora_mes.drop_duplicates(inplace=True)
print("Nulos:\n", df_lesion_hora_mes.isnull().sum())
df_lesion_hora_mes.dropna(inplace=True)


print(df_lesion_hora_mes)



#Tendencias anuales
plt.figure(figsize=(12,6))
df_lesion_hora_mes.groupby("Año")["Cantidad_lesionados"].sum().plot(kind="line", marker="o")
plt.title("Tendencias anuales- Lesionados accidentes")
plt.ylabel("Cantidad de lesionados - por mes")
plt.xlabel("Año")
plt.xticks(rotation=45)
plt.show()



#Configuracion estetica para estos graficos
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"]=(12,6)

#Distribucion general por rango de horario
plt.figure()
df_lesion_hora_mes.groupby("Rango de hora")["Cantidad_lesionados"].sum().plot(kind="barh", color="crimson")
plt.title("Total de lesiones - rango de hora")
plt.xlabel("Cantidad lesionados")
plt.ylabel("Rango hora")
plt.show()

#Proporcion relativa de lesiones por rango de hora

proporcion=df_lesion_hora_mes.groupby("Rango de hora")["Cantidad_lesionados"].sum()/df_lesion_hora_mes["Cantidad_lesionados"].sum()*100
print("Proporcion de lesiones por rango de hora (%):")
print(proporcion)

# Crear un cuadro con la proporcion
plt.figure()
sns.barplot(x=proporcion.values, y=proporcion.index)
plt.title("Proporción de Lesiones por Rango de Hora (%)")
plt.xlabel("Proporción (%)")
plt.ylabel("Rango de Hora")
plt.show()

#Comparacion interanual  por rango horarica (heatmap)

pivot=df_lesion_hora_mes.pivot_table(index="Rango de hora", columns="Año", values="Cantidad_lesionados", aggfunc="sum")
plt.figure(figsize=(14,8))
sns.heatmap(pivot, cmap="Reds", annot=True, fmt="d")
plt.title("Heatmap de lesiones por rango de hora y ano")
plt.show()

#Evolucion temporar de cada rango horario

plt.figure(figsize=(14,6))
sns.lineplot(data=df_lesion_hora_mes, x="Año", y="Cantidad_lesionados", hue="Rango de hora", marker="o", palette="Set2")
plt.title("Evolucion interanual de lesiones por rango horario")
plt.xlabel("Año")
plt.ylabel("Cantidad de lesionados")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

#Variacion dentro de cada rango horario (boxplot)
plt.figure(figsize=(12,8))
sns.boxplot(data=df_lesion_hora_mes, x="Rango de hora", y="Cantidad_lesionados", palette="Set2")
plt.title("Distribucion de lesionados por rango horario (detencion outliers)")
plt.ylabel("Cantidad de lesionados")
plt.xticks(rotation=45)
plt.show()



# 1) Cargar CSV
df = pd.read_csv("cuadro-accidente-año-rango-horas.csv")

# 2) Renombrar columna inicial
df = df.rename(columns={"Anos": "Rango_hora"})

# 3) Convertir columnas de años a numérico (quitando comas)
year_cols = [c for c in df.columns if c != "Rango_hora"]
for c in year_cols:
    df[c] = (
        df[c].astype(str)
             .str.replace(",", "", regex=False)
             .replace({"": np.nan})
             .astype(float)
    )

# 4) Opcional: excluir filas no útiles para análisis por rango
df_clean = df[~df["Rango_hora"].isin(["Total", "No declarada"])].copy()

# 5) Pasar a formato largo
df_long = df_clean.melt(
    id_vars="Rango_hora",
    value_vars=year_cols,
    var_name="Año",
    value_name="Muertes"
)

# 6) Tipos y limpieza final
df_long["Año"] = pd.to_numeric(df_long["Año"], errors="coerce")
df_long = df_long.dropna(subset=["Muertes"]).reset_index(drop=True)

df_long.head()

# Estadísticas por rango horario
stats_by_hour = df_long.groupby("Rango_hora")["Muertes"].agg(["mean", "median", "std", "sum"]).reset_index()

# Estadísticas por año
stats_by_year = df_long.groupby("Año")["Muertes"].agg(["mean", "median", "std", "sum"]).reset_index()

# Top rangos críticos por acumulado
top_hours = stats_by_hour.sort_values("sum", ascending=False).head(5)

# Años anómalos por IQR (picos/caídas)
q1, q3 = stats_by_year["sum"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
anomalous_years = stats_by_year[(stats_by_year["sum"] < lower) | (stats_by_year["sum"] > upper)]

stats_by_hour, stats_by_year, top_hours, anomalous_years


sns.set(style="whitegrid")

# Orden por acumulado
order = df_long.groupby("Rango_hora")["Muertes"].sum().sort_values(ascending=False).index

# 1) Barras: muertes acumuladas por rango horario
plt.figure(figsize=(12,6))
sns.barplot(data=df_long, x="Rango_hora", y="Muertes", estimator=sum, order=order, palette="Reds")
plt.title("Muertes acumuladas por rango horario")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2) Líneas: evolución anual por rango horario
plt.figure(figsize=(14,8))
sns.lineplot(data=df_long, x="Año", y="Muertes", hue="Rango_hora", marker="o", palette="tab10")
plt.title("Evolución anual de muertes por rango horario")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

# 3) Heatmap Año vs Rango horario
pivot = df_long.pivot_table(index="Rango_hora", columns="Año", values="Muertes", aggfunc="sum")
plt.figure(figsize=(14,8))
sns.heatmap(pivot, cmap="Reds", annot=True, fmt=".0f")
plt.title("Heatmap de muertes por año y rango horario")
plt.tight_layout()
plt.show()

# 4) Boxplot: variabilidad por rango horario
plt.figure(figsize=(12,6))
sns.boxplot(data=df_long, x="Rango_hora", y="Muertes", palette="Set2", order=order)
plt.title("Distribución de muertes por rango horario (variabilidad)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()




# Umbrales por percentiles del acumulado total por rango
totales = df_long.groupby("Rango_hora")["Muertes"].sum()
p33, p66 = totales.quantile([0.33, 0.66])

def riesgo_por_rango(r):
    total = totales.loc[r]
    if total <= p33: return "Bajo"
    elif total <= p66: return "Medio"
    else: return "Alto"

df_long["Riesgo_rango_hist"] = df_long["Rango_hora"].apply(riesgo_por_rango)
df_long.head()


# Umbrales por año (percentiles dentro de cada año)
def riesgo_dinamico(grp):
    p33, p66 = grp["Muertes"].quantile([0.33, 0.66])
    return pd.cut(
        grp["Muertes"],
        bins=[-np.inf, p33, p66, np.inf],
        labels=["Bajo", "Medio", "Alto"]
    )

df_long["Riesgo_anual"] = df_long.groupby("Año", group_keys=False).apply(riesgo_dinamico)
df_long.head()

```

```python
# ============================
# 1. Cargar y limpiar datos
# ============================
df = pd.read_csv("cuadro-muertes-accidente-sexo-grupo-edad.csv", encoding="latin1")

# Renombrar columnas
df.columns = ["Año", "Rango_edad", "Hombre", "Mujer"]

# Corregir errores de codificación
df["Rango_edad"] = df["Rango_edad"].str.replace("a�os", "años", regex=False)\
                                   .str.replace("m�s", "más", regex=False)\
                                   .str.strip()

# Convertir a numérico
df["Año"] = pd.to_numeric(df["Año"], errors="coerce")
df["Hombre"] = pd.to_numeric(df["Hombre"], errors="coerce")
df["Mujer"] = pd.to_numeric(df["Mujer"], errors="coerce")
df["Total"] = df["Hombre"] + df["Mujer"]

# ============================
# 2. Estadísticas por año
# ============================
totales = df.groupby("Año")[["Hombre", "Mujer", "Total"]].sum().reset_index()
print("\n📊 Totales por año:\n", totales.head())

# ============================
# 3. Estadísticas por grupo de edad
# ============================
edad_stats = df[df["Rango_edad"] != "No declarada"].groupby("Rango_edad")[["Hombre", "Mujer", "Total"]].sum().reset_index()
edad_stats = edad_stats.sort_values("Total", ascending=False)
print("\n🔥 Grupos de edad con más muertes:\n", edad_stats.head())

# ============================
# 4. Proporción de edad declarada vs no declarada
# ============================
decl = df[df["Rango_edad"] != "No declarada"].groupby("Año")[["Hombre", "Mujer", "Total"]].sum().reset_index()
no_decl = df[df["Rango_edad"] == "No declarada"].groupby("Año")[["Hombre", "Mujer", "Total"]].sum().reset_index()
prop = decl.merge(no_decl, on="Año", suffixes=("_decl", "_no_decl"))
for col in ["Hombre", "Mujer", "Total"]:
    prop[f"{col}_pct_decl"] = prop[f"{col}_decl"] / (prop[f"{col}_decl"] + prop[f"{col}_no_decl"])
print("\n📌 Proporción de edad declarada por año:\n", prop[["Año", "Total_pct_decl"]].head())

# ============================
# 5. Detección de años anómalos (z-score)
# ============================
def detectar_anomalos(df, col):
    mu = df[col].mean()
    sigma = df[col].std()
    df["zscore"] = (df[col] - mu) / sigma
    df["anomalo"] = df["zscore"].abs() > 2
    return df

anom_h = detectar_anomalos(totales.copy(), "Hombre")
anom_m = detectar_anomalos(totales.copy(), "Mujer")
print("\n🚨 Años anómalos (Hombres):\n", anom_h[anom_h["anomalo"]])
print("\n🚨 Años anómalos (Mujeres):\n", anom_m[anom_m["anomalo"]])

# ============================
# 6. Visualizaciones
# ============================
sns.set(style="whitegrid")

# Evolución anual por sexo
plt.figure(figsize=(10,5))
sns.lineplot(data=totales, x="Año", y="Hombre", label="Hombre")
sns.lineplot(data=totales, x="Año", y="Mujer", label="Mujer")
plt.title("Muertes por año y sexo")
plt.tight_layout()
plt.show()

# Barras por grupo de edad (Total)
plt.figure(figsize=(12,6))
sns.barplot(data=edad_stats, x="Rango_edad", y="Total", palette="Purples")
plt.xticks(rotation=45)
plt.title("Muertes acumuladas por grupo de edad")
plt.tight_layout()
plt.show()

# Heatmap por grupo de edad y año (Hombres)
pivot_h = df[df["Rango_edad"] != "No declarada"].pivot_table(index="Rango_edad", columns="Año", values="Hombre", aggfunc="sum")
plt.figure(figsize=(12,6))
sns.heatmap(pivot_h, cmap="Blues", linewidths=0.5, annot=True, fmt="d")
plt.title("Heatmap muertes por grupo de edad y año (Hombres)")
plt.tight_layout()
plt.show()

# Heatmap por grupo de edad y año (Mujeres)
pivot_m = df[df["Rango_edad"] != "No declarada"].pivot_table(index="Rango_edad", columns="Año", values="Mujer", aggfunc="sum")
plt.figure(figsize=(12,6))
sns.heatmap(pivot_m, cmap="Reds", linewidths=0.5, annot=True, fmt="d")
plt.title("Heatmap muertes por grupo de edad y año (Mujeres)")
plt.tight_layout()
plt.show()

# Línea: declaradas vs no declaradas (Total)
plt.figure(figsize=(10,5))
sns.lineplot(data=prop, x="Año", y="Total_decl", label="Declaradas")
sns.lineplot(data=prop, x="Año", y="Total_no_decl", label="No declaradas")
plt.title("Muertes declaradas vs no declaradas por año")
plt.tight_layout()
plt.show()
```
